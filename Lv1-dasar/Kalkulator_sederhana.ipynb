```python
#input angka no 1
a1 = float(input('Masukan angka :'))
#input angka no 2
a2 = float(input('Masukan angka :'))
#ketika a1 dan a2 sudah terisi, input tersebut masuk ke proses di bawah dn menghasil kan Output
print('Hasil dari semua operator Aritmatika')
print('=== + ===')
print(a1,'+',a2,'=',a1+a2)
print('=== - ===')
print(a1,'-',a2,'=',a1-a2)
print('=== * ===')
print(a1,'*',a2,'=',a1*a2)
print('=== / ===')
print(a1,'/',a2,'=',a1/a2)
print('=== % ===')
print(a1,'%',a2,'=',a1%a2)
print('=== ** ===')
print(a1,'**',a2,'=',a1**a2)
print('=== // ===')
print(a1,'//',a2,'=',a1//a2)
```
